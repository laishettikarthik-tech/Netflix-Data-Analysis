```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Task_1_Data_Cleaning/netflix_cleaned.csv")

df.head()
```

```python
type_counts = df["type"].value_counts()

print(type_counts)
```

```python
type_percent = df["type"].value_counts(normalize=True) * 100

print(type_percent)
```

```python
import matplotlib.pyplot as plt

type_counts.plot(
    kind="bar",
    title="Netflix Content Type Distribution"
)

plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    type_counts,
    labels=type_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Netflix Movies vs TV Shows")
plt.show()
```

```python
print("KEY FINDINGS")
print("----------------------------")
print(f"Movies: {type_counts['Movie']} ({type_percent['Movie']:.2f}%)")
print(f"TV Shows: {type_counts['TV Show']} ({type_percent['TV Show']:.2f}%)")
```
