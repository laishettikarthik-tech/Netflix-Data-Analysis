```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv("../Task_1_Data_Cleaning/netflix_cleaned.csv")

df.head()
```

```python
df["release_year"].head(20)
```

```python
yearly_content = df["release_year"].value_counts().sort_index()

yearly_content
```

```python
peak_year = yearly_content.idxmax()
peak_count = yearly_content.max()

print("Peak Release Year:", peak_year)
print("Number of Titles:", peak_count)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    yearly_content.index,
    yearly_content.values,
    marker="o"
)

plt.title("Netflix Content Releases by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
recent_years = yearly_content.loc[2017:2021]

recent_years
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    recent_years.index,
    recent_years.values,
    marker="o"
)

plt.title("Netflix Content Releases: 2017–2021")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
print("KEY BUSINESS INSIGHTS")
print("----------------------------")

print(f"Peak release year: {peak_year}")
print(f"Titles released in peak year: {peak_count}")

print(f"2017 releases: {recent_years.loc[2017]}")
print(f"2021 releases: {recent_years.loc[2021]}")

print("Netflix content releases increased strongly up to 2018.")
print("After 2018, the number of releases shows a declining trend.")
```

## Key Business Insights

- 2018 was the peak release year with 1,146 titles.
- Content releases increased significantly from 2017 to 2018.
- After 2018, the number of releases gradually declined.
- 2021 recorded 592 titles in the dataset.
- The trend chart clearly shows a strong growth phase followed by a decline in recent years.
